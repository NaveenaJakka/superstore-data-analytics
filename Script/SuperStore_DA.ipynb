```python
df.columns=df.columns.str.strip()
```

```python
df=df.dropna(subset=['Order ID', 'Product ID', 'Customer ID'])
```

```python
df=df.drop_duplicates(subset=['Row ID'])
```

```python
df["Order Date"]=pd.to_datetime(df["Order Date"],errors='coerce')
```

```python
df["Ship Date"]=pd.to_datetime(df["Ship Date"],errors='coerce') 
```

```python
df.shape
```

```python
dim_products=df[['Product ID', 'Product Name', 'Category', 'Sub-Category']].drop_duplicates(subset=['Product ID'])
```

```python
dim_products.shape
```

```python
dim_products["Product ID"].duplicated().sum()
```

```python
dim_customer=df[['Customer ID', 'Customer Name', 'Segment', 'Country', 'City', 'State', 'Postal Code']].drop_duplicates(subset=['Customer ID'])
```

```python
dim_customer.duplicated().sum()
```

```python
dates = pd.concat([
    df['Order Date'],
    df['Ship Date']
]).dropna().drop_duplicates()

dim_date = pd.DataFrame({
    'Date': sorted(dates.unique())
})
```

```python
dim_date['Date_Key'] = (
    dim_date['Date'].dt.strftime('%Y%m%d').astype(int)
)

dim_date['Year'] = dim_date['Date'].dt.year

dim_date['Month'] = dim_date['Date'].dt.month

dim_date['Month_Name'] = dim_date['Date'].dt.month_name()

dim_date['Quarter'] = (
    'Q' + dim_date['Date'].dt.quarter.astype(str)
)

dim_date['Day'] = dim_date['Date'].dt.day
```

```python
dim_date.shape
```

```python
dim_date.head()
```

```python
# let's create a fact table that contains the measures we want to analyze. In this case, we will create a fact table that contains the sales and profit for each order. 
fact_sales = df[['Row ID','Order ID','Order Date', 'Ship Date','Ship Mode','Product ID', 'Customer ID','Sales','Quantity','Discount','Profit']].copy()
```

```python
fact_sales.columns=['Row_ID','Order_ID','Order_Date', 'Ship_Date','Ship_Mode','Product_ID', 'Customer_ID','Sales','Quantity','Discount','Profit']
```

```python
fact_sales.columns
```

```python
dim_order=df[['Order ID', 'Order Date', 'Ship Date', 'Ship Mode']].drop_duplicates(subset=['Order ID'])
```

```python
dim_order.shape
```

```python
fact_sales.shape
```

```python
print("Product:", len(dim_products))

print("Customer:", len(dim_customer))

print("Date:", len(dim_date))

print("Order:", len(dim_order))

print("Sales:", len(fact_sales))
```

```python
# Check for primary key uniqueness in the dimension and fact tables
print(dim_products['Product ID'].is_unique)

print(dim_customer['Customer ID'].is_unique)

print(dim_order['Order ID'].is_unique)

print(fact_sales['Row_ID'].is_unique)
```

```python
# Check for foreign key violations between fact_sales and dim_products
# For example, if there are any Product_IDs in fact_sales that do not exist in dim_products, those would be considered foreign key violations. We can check for this using the following code:
invalid_products = fact_sales[
    ~fact_sales['Product_ID'].isin(
        dim_products['Product ID']
    )
]

print(invalid_products)
```

```python
invalid_customers = fact_sales[
    ~fact_sales['Customer_ID'].isin(
        dim_customer['Customer ID']
    )
]

print(invalid_customers)
```

```python
invalid_orders = fact_sales[
    ~fact_sales['Order_ID'].isin(
        dim_order['Order ID']
    )
]

print(invalid_orders)
```

```python
dim_products.to_csv(
    'dim_product.csv', index=False
)

dim_customer.to_csv(
    'dim_customer.csv', index=False
)

dim_date.to_csv(
    'dim_date.csv', index=False
)

dim_order.to_csv(
    'dim_order.csv', index=False
)

fact_sales.to_csv(
   'fact_sales.csv', index=False
)
```
